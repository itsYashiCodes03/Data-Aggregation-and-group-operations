```python
import pandas as pd

data = {    'Order_ID': [101, 102, 103, 104, 105, 106,107, 108, 109, 110, 111, 112],
             'Customer': ['Amit', 'Neha', 'Rahul', 'Priya', 'Amit', 'Neha', 'Karan', 'Priya','Rahul', 'Karan', 'Amit', 'Neha'],
            'City': ['Delhi', 'Mathura', 'Delhi', 'Agra','Mathura', 'Delhi', 'Agra', 'Mathura','Agra', 'Delhi', 'Agra', 'Mathura'],
            'Category': ['Electronics', 'Clothing', 'Furniture','Electronics', 'Clothing', 'Furniture','Electronics', 'Furniture', 'Clothing','Electronics', 'Clothing', 'Furniture'],
            'Product': ['Laptop', 'Shirt', 'Table', 'Mobile','Jeans', 'Chair', 'Headphones', 'Table','Jacket', 'Mobile', 'Shirt', 'Chair'],
            'Quantity': [2, 3, 1, 2, 4, 1, 2, 3, 2, 1, 5, 2],
            'Unit_Price': [50000, 1200, 15000, 20000,1800, 8000, 2500, 15000,2500, 20000, 1200, 8000],
            'Payment_Mode': ['UPI', 'Card', 'Cash', 'UPI','Card', 'UPI', 'Cash', 'Card', 'UPI', 'Cash', 'Card', 'UPI'] }
df = pd.DataFrame(data)
df['Sales'] = df['Quantity'] * df['Unit_Price']
print(df)
```

```python
city_sales=df.groupby('City')['Sales'].sum()
city_sales
```

# **Grouping by multiple columns**

```python
city_category_sales=df.groupby(['City','Category'])['Sales'].sum()
city_category_sales
```

```python
print(city_category_sales.sort_values(ascending=False))
```

# **Group by one and select more than one numerical column**

```python
city_summary=df.groupby('City')[['Sales','Quantity']].sum()
city_summary
```

```python
city_category_table=city_category_sales.reset_index()
city_category_table
```

# **Multiple aggregation with agg()**

```python
sales_report=df.groupby('City')['Sales'].agg(['min','max','sum','mean','count'])
sales_report
```

# **Named aggregation across different columns**

```python
category_report=df.groupby('Category').agg(
    Total_Sales=('Sales','sum'),
    Average_Sales=('Sales','mean'),
    Total_Quantity=('Quantity','sum'),
    No_of_Orders=('Order_ID','count'),
    )
category_report
```

# **Custom aggregation**

```python
def sales_range(values):
  return values.max()-values.min()
```

```python
range_report=df.groupby('Category')['Sales'].agg(sales_range)
range_report
```

```python
range_report2=df.groupby('Category')['Sales'].agg(
    lambda values:values.max()-values.min()
)
range_report2
```

# **Applying operations to groupby with apply()**

```python
top_orders=df.groupby('City',group_keys=False).apply(lambda x:x.nlargest(2,'Sales'))

top_orders
```

# **Group transformation with transform()**

```python
df['City_Avg_Sales']=df.groupby('City')['Sales'].transform('mean')
df
```

```python
import numpy as np
df['Performance']=np.where(
    df['Sales']>df['City_Avg_Sales'],
    'Above Average',
    'At or Below Average'
)
```

```python
print(df[['Order_ID','City','Sales','City_Avg_Sales','Performance']])
```
